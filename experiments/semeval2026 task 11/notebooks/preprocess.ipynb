```python
'''
extract syllogisms from json
transform syllogisms from NL to FOL (chatgpt)
load and add transformed FOL syllogisms to original data
'''
from google.colab import drive
drive.mount('/content/drive')
```

```python
# extract syllogisms from pilot json
import json

with open('/content/drive/MyDrive/Colab Notebooks/SemEval2026/data/pilot_data.json') as json_data:
    pilot_data = json.load(json_data)
    json_data.close()


syllogisms = []
for item in pilot_data:
  syllogisms.append(item['syllogism'])

# write to file
with open('syllogisms_pilot.txt', 'w') as f:
    for line in syllogisms:
        f.write(f"{line}\n")
```

```python
# extract syllogisms from train json
with open('/content/drive/MyDrive/Colab Notebooks/SemEval2026/data/train_data.json') as json_data:
    train_data = json.load(json_data)
    json_data.close()

train_syllogisms = []
for item in train_data:
  train_syllogisms.append(item['syllogism'])

# write to file
with open('syllogisms_train.txt', 'w') as f:
    for line in syllogisms:
        f.write(f"{line}\n")
```

```python
# now incorporate FOL transformations to pilot data
with open('/content/drive/MyDrive/Colab Notebooks/SemEval2026/data/syllogisms_pilot_fol.txt', 'r') as file:
    pilot_fol = file.readlines()

print(pilot_fol[0])
print(pilot_fol[1])
print(len(pilot_fol))
```

```python
# now incorporate FOL transformations to train data
with open('/content/drive/MyDrive/Colab Notebooks/SemEval2026/data/syllogisms_train_fol.txt', 'r') as file:
    train_fol = file.readlines()

print(train_fol[0])
print(train_fol[1])
print(len(train_fol))
```

```python
# add FOL data to pilot
for item, fol_item in zip(pilot_data, pilot_fol):
  item['fol'] = fol_item

# add FOL data to train
for item, fol_item in zip(train_data, train_fol):
  item['fol'] = fol_item
```

```python
# test that fol data has been merged successfully
print("*** PILOT DATA ***")
print(pilot_data[0])
print(pilot_data[1])
print("*** TRAIN DATA ***")
print(train_data[0])
print(train_data[1])
```

```python
# export pilot and train datasets
import json

with open("pilot_fol.json", 'w') as final:
  json.dump(pilot_data, final)

with open("train_fol.json", 'w') as final:
  json.dump(train_data, final)
```
